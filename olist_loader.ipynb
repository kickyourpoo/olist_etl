```python
# create a new table cust_id_dict with distinct values of o_id and o_cust_id columns
query = '''
    CREATE TABLE cust_id_dict AS
    SELECT DISTINCT o_id, o_cust_id
    FROM orders;
'''
engine.execute(query)
```

```python
# join all rows in payments table with custid
query = """
    CREATE TABLE payment_custid AS
    SELECT o.*, c.o_cust_id as p_cust_id
    FROM order_payments as o
    LEFT JOIN cust_id_dict as c
    ON o.op_order_id = c.o_id;
"""
engine.execute(query)
```

```python
# create a new table cust_id_dict with distinct values of o_id and o_cust_id columns
query = '''
    CREATE TABLE orders_items_payments AS
    SELECT *
    FROM payment_custid as p
    FULL OUTER JOIN orders as o
    ON p.p_unique_o_id = o.o_unique_o_id;
'''
engine.execute(query)
```

```python
engine.execute("ALTER TABLE orders_items_payments DROP COLUMN op_index, DROP COLUMN p_unique_o_id, DROP COLUMN o_index, DROP COLUMN o_unique_o_id")
```
