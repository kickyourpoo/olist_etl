Upon viewing the customer_city column, there are dashes/hyphens in a few city names which are correct city names based on internet search eg) 2079, 2098, 2124, 2162.
However, we will remove them to effect ease of joining tables and later data loading

```python
#We will copy the original dataset to begin data cleaning
df1=df.copy()
```

```python
for col in ['customer_city']:
   df1['customer_city'] = df1['customer_city'].str.replace(r'-',' ')
    
```

```python
df1.loc[[2079, 2098, 2124, 2162], 'customer_city']
```

```python
#customer_id = temporary id generated every time the customer places an order
#customer_unique_id = unique id identifying customers, generated at the time of Signup

#View number of orders placed per state
df1.groupby('customer_state')['customer_id'].count()
```

```python
#View number of customers per state
df1.groupby('customer_state')['customer_unique_id'].count()
```

```python
#aggregate by indicator (customer_state)

df1.groupby('customer_state').size()
```

```python
# ALTERNATIVE GROUP BY COUNT

df1.groupby('customer_state')['customer_zip_code_prefix'].agg('count')
```

```python
df1 = df1.rename(columns={'customer_id': 'c_id', 'customer_unique_id': 'c_unique_id',
                                       'customer_zip_code_prefix':'c_zipcode', 'customer_city':'c_city',
                                       'customer_state':'c_state'})
df1
```

```python
df1.to_csv("olist_customers_clean.csv", index=False)
```
