Viewing the Seller_City column, index 360 has an accent character and index 1346 has a backslash which will require an escape sequence. Hence, we will use Unidecode to get normal characters. 

```python
#We will copy the original dataset to begin data cleaning
df1=df.copy()
```

```python
from unidecode import unidecode
unidecode('são paulo')
```

```python
for col in ['seller_city']:
    df1[col] = df1[col].apply(unidecode).str.replace(r'[^\w\s]+','')
```

```python
#Result of indices 360 & 1346, no longer have accent and backslash. 
#Index 1346 may now be cleaned and replaced

df1.loc[[360, 1346], 'seller_city']
```

We will proceed to merge two dataframes with geolocation dataset in order to include latitude and longitude for Olist Sellers Dataset.

```python
#Import zipcode with coordinates dataset in order to merge 2 dataframes to include latitude and longtitude
#rename zipcode prefix column before merge

df2 = pd.read_csv('zip_code_with_coordinates.csv').rename(columns={'geolocation_zip_code_prefix':'zip_code_prefix'})
df2
```

```python
#View seller dataframe again
df1 = pd.read_csv('olist_sellers_dataset.csv').rename(columns={'seller_zip_code_prefix':'zip_code_prefix'})
df1
```

```python
#merge on common field
merged_df = pd.merge(df1, df2, on='zip_code_prefix')
merged_df
```

```python
#rename zipcode prefix to be customer zipcode prefix

merged_df1 = merged_df.rename(columns={'zip_code_prefix':'seller_zip_code_prefix'})

merged_df1
```

```python
#drop unnamed column

merged_df2=merged_df1.drop(['Unnamed: 0'], axis=1)
merged_df2
```

```python
#aggregate by indicator (customer_state)

merged_df2.groupby('seller_state').size()
```

```python
# ALTERNATIVE GROUP BY COUNT

merged_df2.groupby('seller_state')['seller_zip_code_prefix'].agg('count')
```

```python
merged_df2 = merged_df2.rename(columns={'seller_id': 's_id', 
                                       'seller_zip_code_prefix':'s_zipcode', 'seller_city':'s_city',
                                       'seller_state':'s_state', 'geolocation_lat':'s_lat',
                                       'geolocation_lng':'s_lng'})
merged_df2
```

```python
merged_df2.to_csv("olist_sellers_clean.csv", index=False)
```
