### To update the missing value of field (order_delivered_customer_date). Referenece the value from csv file (olist_order_reviews_dataset.csv) and field (review_creation_date). Assumption = take the value of (review_creation_date - 1 day) for the missing value

```python
tmp22 = tmp2[(tmp2['order_status']=='delivered') & (tmp2['order_delivered_customer_date'].isna())]
tmp22
```

```python
NumRec = 0
NoRec = 0
for i in tmp22.index:

    orderid = tmp22.loc[i,['order_id']]
    
    try:
        getdate = chkr.loc[chkr['order_id'] == orderid[0],'review_creation_date'].values[0]
        chgdate   = datetime.strptime(getdate, '%Y-%m-%d %H:%M:%S')
        newdate = chgdate - pd.DateOffset(days=1)
        df.loc[i, ['order_delivered_customer_date']] = str(newdate)
        NumRec += 1

    except Exception:
        NoRec += 1               
    
    
print(str(NumRec) + ' records updated.')
print(str(NoRec) + ' records NOT found in order_reviews file.')
```

# Merging Order_Payments, Orders and Order_Items Dataset

```python
# left merge orders with items
cbm_df = pd.merge(df, chkom, on='order_id', sort=True, how='left')
```

```python
# create ID for orders_items table
cbm_df.insert(0, 'id', range(1, 1 + cbm_df.shape[0]))
```

```python
# rename columns
cbm_df = cbm_df.rename(columns={'id': 'o_index', 
                                'order_id': 'o_id', 
                                'order_item_id': 'o_item_id', 
                                'product_id': 'o_product_id', 
                                'seller_id': 'o_seller_id', 
                                'shipping_limit_date': 'o_tocarrier_limit_date', 
                                'price': 'o_price', 
                                'freight_value': 'o_freight_value', 
                                'customer_id': 'o_cust_id', 
                                'order_status': 'o_status', 
                                'order_purchase_timestamp': 'o_purchase_date', 
                                'order_approved_at': 'o_approve_date', 
                                'order_delivered_carrier_date': 'o_tocarrier_date', 
                                'order_delivered_customer_date': 'o_delivery_date',
                                'order_estimated_delivery_date': 'o_est_delivery_date'})
```

```python
# create new column with unique order_id
cbm_df['o_unique_o_id'] = cbm_df['o_id'] + ' - ' + (cbm_df.groupby('o_id').cumcount() + 1).astype(str)
```

```python
cbm_df.to_csv("olist_orders_clean.csv", index=False)
```
