```python
# create ID
df.insert(0, 'id', range(1, 1 + df.shape[0]))
df
```

```python
# rename columns
df = df.rename(columns={'id': 'op_index', 
                        'order_id': 'op_order_id', 
                        'payment_sequential': 'op_sequence', 
                        'payment_type': 'op_type', 
                        'payment_installments': 'op_installment', 
                        'payment_value': 'op_transact_value'})
```

```python
df['p_count'] = df.groupby('op_order_id')['op_index'].transform('count')
df
```

```python
# create new column with unique order_id
df['p_unique_o_id'] = df['op_order_id'] + ' - ' + (df.groupby('op_order_id').cumcount() + 1).astype(str)
```

```python
df.to_csv("olist_order_payments_clean.csv", index=False)
```
