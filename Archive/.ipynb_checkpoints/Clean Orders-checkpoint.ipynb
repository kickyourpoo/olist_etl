### To update the missing value of field (order_delivered_carrier_date). Referenece the value from csv file (olist_order_items_dataset.csv) and field (shipping_limit_date)

```python
NumRec = 0
NoRec = 0
for i in tmp11.index:

    orderid = tmp11.loc[i,['order_id']]
    
    try:
        getdate = chkom.loc[chkom['order_id'] == orderid[0],'shipping_limit_date'].values[0]
        df.loc[i, ['order_delivered_carrier_date']] = getdate
        NumRec += 1

    except Exception:
        NoRec += 1               
    
    
print(str(NumRec) + ' records updated.')
print(str(NoRec) + ' records NOT found in order_items file.')
```

### To update the missing value of field (order_delivered_customer_date). Referenece the value from csv file (olist_order_reviews_dataset.csv) and field (review_creation_date). Assumption = take the value of (review_creation_date - 1 day) for the missing value

```python
tmp22 = tmp2[(tmp2['order_status']=='delivered') & (tmp2['order_delivered_customer_date'].isna())]
tmp22
```

```python
NumRec = 0
NoRec = 0
for i in tmp22.index:

    orderid = tmp22.loc[i,['order_id']]
    
    try:
        getdate = chkr.loc[chkr['order_id'] == orderid[0],'review_creation_date'].values[0]
        chgdate   = datetime.strptime(getdate, '%Y-%m-%d %H:%M:%S')
        newdate = chgdate - pd.DateOffset(days=1)
        df.loc[i, ['order_delivered_customer_date']] = str(newdate)
        NumRec += 1

    except Exception:
        NoRec += 1               
    
    
print(str(NumRec) + ' records updated.')
print(str(NoRec) + ' records NOT found in order_reviews file.')
```
